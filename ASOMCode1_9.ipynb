```python
import base64
import math
import random
import time
import zlib
from dataclasses import dataclass

import numpy as np


# ======================================
# Embedded datasets from the paper
# Source: https://cs.joensuu.fi/sipu/datasets/
# Paper: https://doi.org/10.3390/math11091995
# ======================================

AGGREGATION_DATA_B64 = (
    "eNpVWkmSJCkMPEebzVfS2An+/7EJubsk6tDVeAJCOwKizt+cT3t/6/v73786fudpG78BDXR2hxXwBeq/DVSmYAccgk87GI4u+/GjKrSetgSaEdu+QPsNQ6+jDoIC7WKz26/zV9V1pqHj007SryC21QaHalMU72lgkEsVCLbVV36XCg4YCCGL5rV5TVyuvWoUlstWTexpvAEcA8U5/rqGI/SN4L/awOasQNDqy5VL7oOFlpECSL6OTd+02qY07HhBb0ua9wdQBAol46QXpEWhh6rw05K9dmid/rNsoC8iAKlfrQB9LLEIe071jJnaeTF+uFwLYF/6GKlhm9GlYnZ20Tyg2cOMLQbStAL0xAE7GgJXVM8iv1vgMws4mlQBbUBxJhU1TeRFH12ixN/RbprRU6BhE13uYTy6SjB/kO5yqYPFEcQGehrnfwOw4DWiS9iu5QB8ve/HJvmmusTwSUVuqXXL/gBNXQtIEptKWvBfDUhjM9oQvHFGN/GqvGOAVnVRzOOLjAe9FmdtWTuYtiknNZZgglp1s87s2xzo0u1clvIUuSgEqsdtgHUlacGw5satR1Ng3A+VYK966Fk8fD2VgfP9LPf7OKz78kWbQC9ttoaHBCaPBO9P7Y0pL5285Hxkh6qcZqvVLSYRbXVHiil/0CfNknpfyqygoXn6Da6e5vzQCcQPwfrT1TKs65/IbSResWLzPaHKgYdnzpJ+yg3DXZjpt3mypPGbJ9kCgvX3FxVP8iHbQU+h3phii6eiNtPU5/aOivEl81IPEsYjllrax2osVZuvJdm+NueDsHMKl9KSZt7lHctQSDsM9QTHN6kmt3pj//36fEn436t9tcoXYluiZ/jQdnV2DR1ZGSQzX+q4GO3WSHbAGu1u25z7a7UsUXeSgMfWi7x2OvG8uLTtctb/bG0cLEu27TzfSPz+SWvtyjHGt1UMG8Fv4xvGL1JqbK+HRQxAsxkTAOEOuwGAcBP4hlUSxo5N193KrVZykFq19kYPpnPQFmNDCOxPsjxn8L9F69UwEDvi8xhwYb5hhdyYquChm6ndfOr6vV3rIxtsT0XlJ7CwphRYQxbzKTjbVlYq0tKJtc9N6PsfenmGpxpMGAr6WtBEaja+hDbQ6+PqA88a7q+WU4cXfhaPwyOrFh/YqJoNYOUopBmocMuUPQipXYEuixjadJsBV30f1HACZh60lxxCoM1cFRs7kxJpv3KQ4WU3q8kR/l58ZYRsEcPdKKSYw0B3BTxIskNl5OsEkEO2GDFCy1eyTLR9DlwxJnVQOA7ac5LbKk+RuR7LCsOT9ZKW4JexklX6WzIwZKQ/FPMOGHO5Khhql1ZOWMbFHrS7/GiSt+aIelXnkjO+DvdkxKS53O0mvbMmmROGrKlzVHXhj7DxKwqDXfsGwZZF2eVlrj/UXNstYD0rdUFtCpR5dwG1jJ4l9j5Rn5mhs56ZVvtUMV0xSKkmsltqZhw9M+PGBh1n55laxTxlhmmNZHiAtfe14vD1bZHhDoXw6+nuy5Aveo2zRUd6q01KtXUPLZva3STmqyNiHQKU0K6hK/DX5UQznBWnV+/h+ViKsiNKd3IQIH2i/3zCvta39sxBLVzKtsqghEJJBrDtoidfn/w+IzULM86cTsumJNNVNn9XQJpZl2vWvMdNu0jg6hkZQLCu57dzmRf7h3uBHRmHJGC7+ijju+cMt/Om0oqvs24VWHWYFFqqChVHdfF4hHU3Ql/PkSs9zK4YujIOioYukVCfu+uw3g5kLOj3Exy0wkGvet6ne5BjepdOmSK0ygufXnfHdsY2lDiDm1D9RsAN1+KMsNyM/uLDhvmxa/FZHv02fDtluP32VW2c51ycs+TJKGFdaTpMiVxDyLpyWguxUeDL+b9BUMEOAhH/DZWAPKFZmhoUB2teGtiXZ6MmlmeDz+nrF0ZAajEB6s6ZLvKpx1umtpUK3el9pqqR9tny2Jd5u7uq30uJlsIjwUPRbxAvvm9ZxeVbAkwWG+xiKbPc45+ThFv2WPwc+ZRtIL4P2c4/XUQD9Wr3cMjYZ6EVSc9TZuwm5sQ7lIyefnmA3M4ONGbA161OcxJ1RvZf9Apg4NCs9P2mPQgh0pgwJJJdRYQXNO42dJwedobfzXS1N3ZEnq88t7caUcCOHT4UO/f5XXv6y/KiXAF6UtXLi1ZeFpT0jT+VxnvVonZyz5JE5bSnTdaInh/P5ImCA7tXmS+PBsxKtWa+wvBypZ9cx4x/MlzqcyLG4XjnMngUQSzeW6Cm006OrOFCK4tbxHAIjNNucGmmuUZW1emXT0VlaVcPyQwuIs498GRmsuK0XUZ93REKK7IrA71J4E2edtRtvCba7i4tNkzcaixfE6lvRf6jJ7mHRo7gddimCeGiiv5mR4k3iXUTpyWtkwTGpZEu8xwhq4Jf7zpxLmid+qkC+3l92J9CtA2msJNhtpPem7muqfo8SV2bjIHICZbPIycPqsvafTAFfP8+4PmAqCE1WAuebrsP2kWpwVDkegKEcyVAScyfK5L2124M4HqNf9V0mbiGqiGCrW0CTPEcMxx9kqilIsd/HuKdO6GYZ/H5WyntdGaoLdEdntYITHrJiCFD66BUyGENdYS1F9TdNGOqaCM6qgk4R6cAcmBlX7BzuA14z/Y58miXtDNPYhj8WQa79Fy157nWcXTjog1eetwcz9GcdgC8YygREdGdnfbLcCc3A6TF9OROS0F1ziXYMvtAfdOkGiuJXGXpDAvu420LWjaRFjgaApVrSAi7kFxrUD1huZlK6JMb+VCXTXEr8EDJtm4xKOlk0nYj8GTLcZ33OFTP4h0LNfcHDG4v0imuayIulJZjYHPxKlJ9GpmHNy3luwyZOn6rEWGAKw+iNrVjRPAkrLxd2dcqLeNv4yavStY6eTUn9OjwF30tFN51mQOkHVg+gJsnjwec7WmVdTmabggypHgpQ4iNa4Yx75Woy0AUJiDlnjfsrkEUzjyzEj18dc0+I74QeDzOrdA7eCdcdiManWPqFn+FUXhMCN1aSbs87+FQJaR3AiBe9y4PbpxBlocqq2HCogtmovkX2X1vF+AN+jDUrKZlTU/EJ585cyhPV0vOqmqUjB2xwsvCEyzz5nm6qGRsqY839jVVhtqY6KV8hF1PJMOhqQXNoVtwzRt8iyk+7sV6Q0hq8c4GeGJFMOOopy7Q99WjYZ8GbdwwTFv1yhW6wEsDVeYvNlTvyT5skNfA7g8ly7fe1i6KTd5R8j1H8/j458ZkcU/U/S2QEK+vb9jSHi0vkiMlGHprDIMF0d7ETEmfjyWo3dZcTz3el5YSrGsC+SM1iDfIEhbDe4wsWy/PHbKsRJh6gnOEN41wCXrk9s4SNKe/GbFrxevS0saR/s+LpuK6HnyzlHT+dMVp+xKOq9Wfz1qXwvBMXJ0vXHey2BUZvsAFK03Km/nkK6ZhS5ec5tMCfDbO1VoY4NUrdNqqZ+zdRu90pJPRpuIzoC83+RK+BSoe4nuqa8Ti+2Glu3xTasNDG++WIxxu3H1VX1+0Ky7u7DQitl6+1UcgzJ87Pr6RiDw5o0vh6ZmpzWsc2VgutX9kcq55OyOGH6S8Dl9+yRK5Hh9/XGrfmRzx6Y1nw8pvSSId+YcmsdsE510fh0TSK/wG5go8z5VVn/qEQ+w05Zr6kmR5bdRWumJsZoMfTzBv6WH1y5x+hYRjYdXptKFMqX6Rx8Ol3/Xg/Fb9DN+4g+pk/B3Z2JyeRKpuQoJIK0wiQlWfFAyHm+lGfYwlAH76UOMGzRcmweY88kmWb6UuDvad4H9IOH2g0JOxJWcUHX48kypCUqh+xYKvqMgOPkJalwRdqq2M5OXT9GHNhbQcvywZrgYPUDHa+XlLczRyhcYVmkAP8/ALARx7ddkD1tjzztRyU4qvPnCmpgenHYHNPazqiHoPU1oiiaMPDSpfc3kjhU83dMsMgIvPbm18J2NnU33jY/eO/pWNXdeqjUw39SqMBSbfjr1tN1U9GnaR5d+iiANeeu9kQIPUnupYvGeb/PRkkIGjVxcywEs6rT+Cl6G7PNLEsXbqK5qpyRsnz+NswtSTT9CD6uDdwU8/L976c2ZyuXFHTh50Yb8FJm+RJx+ZfQIOmJDlfzW4Hz0="
)

R15_DATA_B64 = (
    "eNpdWluyZKsK/K4TcafSoYKv+U/smplAVfdP90Z5w0LEun9OG5/e/nR7//3vv/eXzc/9s9cJsN3z4HsIvw0D1AOYDlo3gccWNltyGgbSs5a2pxHK3f6IHzy0aY8WrGYg2z6AR0rq1HIsF7xM2y3Ump1aipqi14+e1LKd3OwXm9sLlp6jTF4hLD1ynS7wZH7I0Dx90ihsB3gbTZnnlEex32/u0/DdwpR9iW670A8XVssF77SlHLep3ryn9KE9PVx5Tepb6asQ+lef8Qk0hs3pSyv7Z1dYS2Gn/RYKX7JvI8W1I3XLnYfsb0qn9WMSHH+agvGEDcFH7J4MLXC/PyKAnZkHmSu2jbr0JB/jhO9zm9r4Kf5U3uYJfluJ1p/6QcF8v7k//J/9sbnw7An8LQtKhZfXvRUymRXvJWt2au+T6vWUdhmKdUZqt+m74QFPfZ2tpbBO+70I1tCHcFJCW0BYvfzH5PwKkPkjremK9W0l8UYuaX85v42e0emwqP9YP8ZHOSzy4xC3LNkdZUebFQ9+fEBLeXJnRXBEdpaLmL1rl4Qg6N8AKT3TA10Bg+9jQTlnJ1kavqD+x19lMubopVHwqhGhHS7M96VqYRgpEPtYILw8GSzCeyaDTgaoMVqYQ37+iujiuIoj/bpyf6H8yNpUgRxfvAV3p4hX4wJuIliJME37CaOWdgZcOrMAPo4tRaIiQ/LNBZeOSMf006FfbOQCitxghUlH0KpisaYWyhF9k+JrqC3xTIShDNsjDZ9Sa+6K1vZfS95fCoYVRaeIWVpOhrcMPRDxPsZWSl1m6al4L1HcUkqG71F2Nco8J73rQzlULExa7Z6u4af3KCs+cn+5DjURvnwIjsI6jrz/7CZMzzGsnscISswN9KWUnDPgqxQ8L00It80Ee5FxVXmxsxn7E+Lex+J/sxvBzqasCWqeLp2heAvnJcKiNl3YbrSlBzgveSPnxXwpV5O34lHbm5mKWihLrnxtKy2jrFaqSfMxYp81Z+CDkR+3DJvBzsSul5v/8vJRpJcldvgh/YJ4snBHUBTkFV5QctsOLzWJHi0tCbtHwFPfLIqsmCurZlrqdPr9tbuz1gQ3IQdx1ZwM+IgEWBkzESwGZf85mzn/yCd07zqr5gss4aGT4QmLbXUlN9Db/EQNmWB2lU3dYvfozAxiV8HFWS9YNXyekGUzKnRQqyXa3cX8qEc4NkJYO1lUtC9Vzw72OhNR9kMZkuMIEby039Py/o9pTvx9/9Lm7ESPUm4hbelswLKkbykfyvFDetI8ya+UT2nzROUPY68LLm11es0d7GdzHUWl/fnEeSJnqDoi2JIva+9I/grkTl/KmMRmQe9sq75ZgVobxkZzusvVHpFO42O/skjOsl55guMftWdJ2fU5bB8WpbH9aV0Q4nT58RD0Dtz1UoXgKxKH/ZRwJzbdXHytoScYK/k+Hx121NxFd/Q+Lo9d1MTDywNBHEgPXMmZHY5ZiDVquEbobyINEE39k7NDY91CegtOrywcdlqCJkhPmt7YOMHv4oSeH582N/FRvn+n9i5bqJ5SgjL5Lurbd3hi08PnBicUyae/hZhLP/XrGQ7IOfLE5vdb1j3aBc5ejmFnPm7IbexM0fuWeQhlsDrz14lOsbZyk3JO+gkRv2y5ViYwqlk4ajjjIWScNKB9NW3H3Q3m9gvwGeaAcAXYOnbGZ/NCKvAjizfdxhSZT+ut++hbf8l2BOJ0R7SFK3178sE3sVkKCaJQICFDKA7kzc6J4BigbS3kMAlw0hPaYITCor0lRrlp2C0NUTxRWYMvbjhyjhRuUHiHH5ybqRH63Lcn9Q87hMPTQgrejwq2IPjnaRK2dXx0OGikAqS9hiadxFDg5iR9D53/9RKRl38VPMz3n0j19PcKL3mEddDF8HtIAqsdnKV/km6o85RKHUl5e/C9MMczGJMOj4RA9+ZMpsfhCGY2IYKHaix+H7hqnOzJYXCCq/MzLfJJV97cngrgKXRjyFbCXSUsZF06CK12yGK8txc8megj0Lk7eopGW3nYMp1siVHoPJkbU6BUi2i0hPFto9YlbCxAfafsDT9ZsEtDWwofHduzl3C6ERl26o59lMBhCvHDEe/0oF+KO3VDH3aqWUdP8HWb3Jj7VL10sSlZhX2oupffjKfKSmELPuGtOd2uLzRVU5mwdKQz+VaZgi4gionEq4T6X56x0G4xoTTNuoS7Rk+GCiiEGHz1wvioTRTkMU7ZWpgvsGp9VqJzPoJrShB0zR9O4mtYk/ziuC7+g0cNyofgqeMdp4sWTK0Xup2g2IcUZQCPozW/ErB9vbbFcOX+PTHd+jLUPCZV0OmwUyGexzjkAG/6/8EjqR0DGXR2AqOZKmY4gjS+COY8a+ZIbbpxUpjoa+dgLLgfdj+og6HOjqll7E9Z1ytecvdZ5U+qO/0nui+E6SrOUtJU5ygGX0nKomdH0jq7DU/Zs0lUBZ6DsqD2Z/nzs7+rBxzCFYxBsWK5gJb0pQgvMVgxjgbexuUUiziY8T4cXi2Esw0LfZxcQP/wMM/6riCL/EWtFjoX7pcLu1jXJTfUW5K0V8neXMHFLjW+NOGWLrJg9LKAEwzn91RmUzsr7cIzGqNyZTrVO7++ouRe6soATgWEYbJxna9jwurxNaDRyDNKElovKNwKp8kRdryCIBvP+YfN3SUKo67Hxr6OaFxo/jW70zW2SpKC/TXx/rVgLCJvwW19IysxX/3vUKzbN2UU/R/PoLD/E+yOhTlqwaWufWWbK2Ok/zvZgIzPG9cT9pv2yQ+SMNpUfOA7F3CePp6HUy92fIjz0t1TC/DxYjH55dnW+mWxOEjSAvWB1MSAJUtTXnG4RJjFEqUCZbcL4+oji0rBBabz4lkdailSuvSSBxTaMUHUjRQUyMaeDR04jPW3FhooEUP+fcd2cGzi8LXUtZAItpdEJAc8uNCbyVGeYH3tda2FM3OBN8UVQ0ZhSKuE8YzCiBXCoB2a1zCmU+6eZQcDZBXR/o+W4Ql+0d+seL4c6UvlwCqZnC0uzU5/8qyVoV1pdIslP/QywxXrIYcMLCDh7A9nVv/7zzQuGHoI4/7QrIyzIC7wDcN4FolknxgrJQniMngJF4KGb7ulDE5jjH1GsDQpxZkhhXZiNEt4EmyjCJwLqdPStJGpJwSqtG+qgPsyCHpRdFGMUmoCxhHK/SYOBTsRwxtEoCy9RklHlxvuKBGAyZcqaYCKlyfCU/LLaQoMhosSqKGkr+SPtxU4yXLBNGjcheHyArNBOk7yuEUyFOxeRky6ZdfCIIdVXjkWw8pSc5DCM/aajZZXOGd+Es6q9HFZLYrJs8l0Isvvn8VrQzdCvAHxgDGS4wLJW6lgjhU0BSU+Zm9Ljy6if5m0JE34i/s7tjev2D2ZT9x42MMSxgRDx50pwqDlw6rgAXQ8dBDEWPjxqu2B65/efAW/z2LxkVfUHZbw8VbwBnVL6um8F86EUcW3ZoPUfJG739g+vEb2stNpSvf02+YNqhe7RvzxNbXRtpF+lDi3MgbTJzVU4UfaPpP/mfRr+YKXHg92k9/Z5pAhPY0or4yCnR/bsMugrwraoCetgqwksJ84bQwCwjTctvtIzfAqtPX+GxlEzV3MdVZEzfI4E01PLq658vkYoyGYb2rssQV3DMv/WO4eMFMLSWYT1GMnt3NAzZmouOO9jF9LkD+IDycE4ZdsQHjWQrMEGkWtQMVZYhzAC8RzFC8SAeMJjL9e8Bx7mt70w2bAs+QOYs+vltDLQi+N7p0TtCDfoC70iccy9SsSTtHJnBbjGhuy8H6EMWWwogvwfp6aAsZgI3jhlexFa1Q45HDtb73DvZj7T7hQKtePLvzpQbw5/B17XJSN9/s0jG5sCePmFZ2l8PFwqN42npXgl/63I04oh77wUhs0zXUlmCzCM2452GdyaQHDWFf7NdWxYg6v3zrMaIPjOhMcPlNPlLNa3snHMMHHka0sYbEwP/q4k2CL4Eel9RbURoTOZFEwzuqp00oLeNGZOv0CowGjfWUcwrNMAD6aV4Foypw/2xC64W1Lb3KhAV5C3idZKqCsuX4EJBJnUn1ZvFN78v00Ybxe6PAPDhCBUyG8ghctvsvOurLMuGCWjpPjwKDAWMz5uZcIOrIo+DL5PrSywvCi92tWo1kY+CTGBsX5msn0Ot9QhGNKJCMx+jc3+sf1pB0SxttXf/6///4PIbHzcA=="
)

PAPER_REFERENCE_RESULTS = {
    "aggregation": {
        "CSOM": {"average_mqe": 0.095711, "final_mqe": 0.036482, "time_taken": 1.144511},
        "SOMOC1": {"average_mqe": 0.091741, "final_mqe": 0.034899, "time_taken": 1.154533},
        "SOMOC2": {"average_mqe": 0.070850, "final_mqe": 0.031247, "time_taken": 1.143495},
    },
    "r15": {
        "CSOM": {"average_mqe": 0.113570, "final_mqe": 0.028119, "time_taken": 1.019185},
        "SOMOC1": {"average_mqe": 0.105027, "final_mqe": 0.027102, "time_taken": 1.027379},
        "SOMOC2": {"average_mqe": 0.080154, "final_mqe": 0.021832, "time_taken": 1.018741},
    },
}

PAPER_CONFIGS = {
    "aggregation": {"map_size": (12, 12), "epochs": 1000},
    "r15": {"map_size": (11, 11), "epochs": 1000},
}

K_VALUES = [100, 500, 1000, 5000]
ALGORITHMS = ("CSOM", "SOMOC1", "SOMOC2")
PLOT_STYLES = {
    "CSOM": {"color": "#1f77b4", "linestyle": "-", "marker": None},
    "SOMOC1": {"color": "#d62728", "linestyle": "--", "marker": None},
    "SOMOC2": {"color": "#2ca02c", "linestyle": "-.", "marker": None},
}


def vector_sub(a, b):
    return [a_i - b_i for a_i, b_i in zip(a, b)]


def vector_add(a, b):
    return [a_i + b_i for a_i, b_i in zip(a, b)]


def vector_scale(scalar, vec):
    return [scalar * value for value in vec]


def dot(a, b):
    return sum(a_i * b_i for a_i, b_i in zip(a, b))


def norm(vec):
    return math.sqrt(sum(value * value for value in vec))


def mean(values):
    return sum(values) / len(values) if values else 0.0


def deep_copy_matrix(matrix):
    return [row[:] for row in matrix]


def _as_float_array(values):
    return np.asarray(values, dtype=np.float64)


def _decode_embedded_dataset(encoded_text):
    encoded_text = encoded_text.strip()
    padding = (-len(encoded_text)) % 4
    if padding:
        encoded_text += "=" * padding
    raw = zlib.decompress(base64.b64decode(encoded_text)).decode("utf-8")
    features = []
    labels = []
    for line in raw.strip().splitlines():
        parts = line.split()
        features.append([float(parts[0]), float(parts[1])])
        labels.append(int(float(parts[2])))
    return features, labels


def minmax_normalize(data):
    data_array = _as_float_array(data)
    mins = data_array.min(axis=0)
    maxs = data_array.max(axis=0)
    spans = maxs - mins
    normalized = np.zeros_like(data_array)
    np.divide(data_array - mins, spans, out=normalized, where=spans != 0.0)
    return normalized.tolist()


def load_paper_dataset(name, normalize=True):
    key = name.strip().lower()
    if key == "aggregation":
        features, labels = _decode_embedded_dataset(AGGREGATION_DATA_B64)
    elif key == "r15":
        features, labels = _decode_embedded_dataset(R15_DATA_B64)
    else:
        raise ValueError("dataset name must be 'aggregation' or 'r15'")
    if normalize:
        features = minmax_normalize(features)
    return features, labels


def build_hex_grid(grid_size):
    rows, cols = grid_size
    row_indices = np.repeat(np.arange(rows, dtype=np.float64), cols)
    col_indices = np.tile(np.arange(cols, dtype=np.float64), rows)
    coords = np.empty((rows * cols, 2), dtype=np.float64)
    coords[:, 0] = col_indices + 0.5 * np.mod(row_indices, 2.0)
    coords[:, 1] = (math.sqrt(3.0) / 2.0) * row_indices
    return coords.tolist()


def find_bmu(x_t, weights):
    weights_array = _as_float_array(weights)
    x_array = _as_float_array(x_t)
    diff = weights_array - x_array
    distances = np.einsum("ij,ij->i", diff, diff)
    return int(np.argmin(distances))


def node_distances(coords, bmu_index):
    coords_array = _as_float_array(coords)
    center = coords_array[bmu_index]
    diff = coords_array - center
    return np.sqrt(np.einsum("ij,ij->i", diff, diff)).tolist()


def alpha_t(alpha0, t, total_epochs):
    return alpha0 * math.exp(-float(t) / float(total_epochs))


def sigma_t(sigma0, t, total_epochs):
    return sigma0 * math.exp(-float(t) / float(total_epochs))


def neighborhood(distances, sigma):
    sigma = max(float(sigma), 1e-12)
    distances_array = _as_float_array(distances)
    return np.exp(-(distances_array * distances_array) / (2.0 * sigma * sigma)).tolist()


def mean_quantization_error(data, weights):
    qe_values, _ = _state_metrics(_as_float_array(data), _as_float_array(weights)[None, :, :])
    return float(qe_values[0])


def mqe_gradients(data, weights):
    _, gradients = _state_metrics(_as_float_array(data), _as_float_array(weights)[None, :, :])
    return gradients[0].tolist()


def build_random_matrix(rows, cols, rng):
    return [[rng.random() for _ in range(cols)] for _ in range(rows)]


def _state_metrics(data_array, states_array, batch_size=128):
    n_states = states_array.shape[0]
    n_samples = data_array.shape[0]
    input_dim = data_array.shape[1]
    qe_values = np.empty(n_states, dtype=np.float64)
    gradients = np.zeros((n_states, states_array.shape[1], input_dim), dtype=np.float64)
    sample_sq = np.einsum("nd,nd->n", data_array, data_array)

    for start in range(0, n_states, batch_size):
        stop = min(start + batch_size, n_states)
        state_batch = states_array[start:stop]
        weight_sq = np.einsum("bmd,bmd->bm", state_batch, state_batch)
        cross = np.einsum("nd,bmd->nbm", data_array, state_batch)
        distances_sq = weight_sq[None, :, :] + sample_sq[:, None, None] - (2.0 * cross)
        np.maximum(distances_sq, 0.0, out=distances_sq)
        bmus = np.argmin(distances_sq, axis=2).T
        min_distances_sq = np.take_along_axis(distances_sq.transpose(1, 0, 2), bmus[:, :, None], axis=2)[:, :, 0]
        qe_values[start:stop] = np.sqrt(min_distances_sq).mean(axis=1)

        selected_weights = state_batch[np.arange(stop - start)[:, None], bmus]
        diff = data_array[None, :, :] - selected_weights
        diff_norm = np.linalg.norm(diff, axis=2)
        unit = np.zeros_like(diff)
        np.divide(diff, diff_norm[:, :, None], out=unit, where=diff_norm[:, :, None] > 1e-12)
        batch_gradients = np.zeros((stop - start, state_batch.shape[1], input_dim), dtype=np.float64)
        batch_indices = np.broadcast_to(np.arange(stop - start)[:, None], bmus.shape)
        np.add.at(batch_gradients, (batch_indices, bmus), unit)
        gradients[start:stop] = batch_gradients / float(n_samples)

    return qe_values, gradients


def _backward_costate_step(lambda_next, gradient, beta):
    beta_array = np.asarray(beta, dtype=np.float64)
    gradient_array = np.asarray(gradient, dtype=np.float64)
    lambda_next_array = np.asarray(lambda_next, dtype=np.float64)
    decay = np.exp(-beta_array)
    integral = np.ones_like(beta_array, dtype=np.float64)
    stable_mask = np.abs(beta_array) > 1e-8
    integral[stable_mask] = (-np.expm1(-beta_array[stable_mask])) / beta_array[stable_mask]
    small_beta = beta_array[~stable_mask]
    if small_beta.size:
        integral[~stable_mask] = 1.0 - (small_beta / 2.0) + ((small_beta ** 2) / 6.0)
    return decay * lambda_next_array - integral * gradient_array


@dataclass
class RunHistory:
    weights: list
    bmus: list
    samples: list
    quantization_error: list
    alpha_controls: list
    sigma_controls: list
    radius_history: list
    switching_function_alpha: list
    switching_function_sigma: list
    raw_switching_function_alpha: list
    raw_switching_function_sigma: list
    smoothed_switching_function_alpha: list
    smoothed_switching_function_sigma: list
    sigma_target_history: list
    sigma_delta_history: list
    costates: list
    elapsed_seconds: float

    @property
    def controls(self):
        return self.alpha_controls

    @property
    def radius_controls(self):
        return self.radius_history

    @property
    def switching_function(self):
        return self.switching_function_alpha


class ConventionalSOM:
    def __init__(self, grid_size, input_dim, epochs=1000, alpha0=0.5, sigma0=5.0):
        self.grid_size = tuple(grid_size)
        self.input_dim = int(input_dim)
        self.epochs = int(epochs)
        self.alpha0 = float(alpha0)
        self.sigma0 = float(sigma0)
        self.coords = _as_float_array(build_hex_grid(self.grid_size))
        coord_diff = self.coords[:, None, :] - self.coords[None, :, :]
        self._coord_distances = np.sqrt(np.einsum("ijk,ijk->ij", coord_diff, coord_diff))

    def train(self, data, sample_indices, initial_weights):
        start_time = time.perf_counter()
        data_array = _as_float_array(data)
        weights = _as_float_array(initial_weights).copy()
        sampled_points = []
        bmu_history = []
        weight_history = [weights.copy()]
        alpha_history = []
        sigma_history = []

        for t, sample_index in enumerate(sample_indices):
            x_t = data_array[sample_index]
            diff = weights - x_t
            bmu = int(np.argmin(np.einsum("ij,ij->i", diff, diff)))
            lr = alpha_t(self.alpha0, t, self.epochs)
            radius = sigma_t(self.sigma0, t, self.epochs)
            h_t = np.exp(-(self._coord_distances[bmu] ** 2) / (2.0 * max(radius, 1e-12) ** 2))
            weights += (lr * h_t)[:, None] * (x_t - weights)

            sampled_points.append(x_t.copy())
            bmu_history.append(bmu)
            weight_history.append(weights.copy())
            alpha_history.append(lr)
            sigma_history.append(radius)

        states_array = np.stack(weight_history[1:], axis=0)
        qe_history, _ = _state_metrics(data_array, states_array)
        elapsed_seconds = time.perf_counter() - start_time
        return RunHistory(
            [state.tolist() for state in weight_history],
            bmu_history,
            [sample.tolist() for sample in sampled_points],
            qe_history.tolist(),
            alpha_history,
            sigma_history,
            sigma_history,
            [],
            [],
            [],
            [],
            [],
            [],
            sigma_history,
            [0.0 for _ in sigma_history],
            [],
            elapsed_seconds,
        )


class AdaptiveSOMOptimalControl:
    def __init__(
        self,
        grid_size,
        input_dim,
        epochs=1000,
        alpha_high=0.5,
        alpha_low=0.01,
        sigma_high=5.0,
        sigma_low=1.0,
        policy_passes=1,
        phi_dead_zone=1e-3,
        phi_smoothing=0.25,
        alpha_sharpness=10.0,
        alpha_momentum=0.8,
        sigma_sharpness=10.0,
        sigma_momentum=0.8,
        sigma_control_strength=0.35,
        sigma_step_clip=0.2,
        sigma_phi_dead_zone=1e-5,
        sigma_phi_smoothing=0.1,
        sigma_phi_gain=2.5,
    ):
        self.grid_size = tuple(grid_size)
        self.input_dim = int(input_dim)
        self.epochs = int(epochs)
        self.alpha_high = float(alpha_high)
        self.alpha_low = float(alpha_low)
        self.sigma_high = float(sigma_high)
        self.sigma_low = float(sigma_low)
        self.policy_passes = int(policy_passes)
        self.phi_dead_zone = float(phi_dead_zone)
        self.phi_smoothing = float(phi_smoothing)
        self.alpha_sharpness = float(alpha_sharpness)
        self.alpha_momentum = float(alpha_momentum)
        self.sigma_sharpness = float(sigma_sharpness)
        self.sigma_momentum = float(sigma_momentum)
        self.sigma_control_strength = float(sigma_control_strength)
        self.sigma_step_clip = float(sigma_step_clip)
        self.sigma_phi_dead_zone = float(sigma_phi_dead_zone)
        self.sigma_phi_smoothing = float(sigma_phi_smoothing)
        self.sigma_phi_gain = float(sigma_phi_gain)
        self.coords = _as_float_array(build_hex_grid(self.grid_size))
        coord_diff = self.coords[:, None, :] - self.coords[None, :, :]
        self._coord_distances = np.sqrt(np.einsum("ijk,ijk->ij", coord_diff, coord_diff))

    def _smooth_phi(self, phi, smoothing=None):
        phi_array = np.asarray(phi, dtype=np.float64)
        if phi_array.size == 0:
            return phi_array
        smoothed = np.empty_like(phi_array)
        smoothing = self.phi_smoothing if smoothing is None else smoothing
        smoothing = min(max(smoothing, 0.0), 1.0)
        smoothed[0] = phi_array[0]
        for t in range(1, len(phi_array)):
            smoothed[t] = ((1.0 - smoothing) * smoothed[t - 1]) + (smoothing * phi_array[t])
        return smoothed

    def _apply_phi_dead_zone(self, phi, threshold=None):
        phi_array = np.asarray(phi, dtype=np.float64).copy()
        threshold = self.phi_dead_zone if threshold is None else threshold
        phi_array[np.abs(phi_array) < threshold] = 0.0
        return phi_array

    def _prepare_phi(self, phi, smoothing, dead_zone, gain=1.0):
        raw_phi = np.asarray(phi, dtype=np.float64)
        smoothed_phi = self._smooth_phi(raw_phi, smoothing=smoothing)
        filtered_phi = self._apply_phi_dead_zone(smoothed_phi, threshold=dead_zone)
        scale = max(np.percentile(np.abs(filtered_phi), 75), 1e-8)
        effective_phi = gain * (filtered_phi / scale)
        effective_phi = np.clip(effective_phi, -60.0, 60.0)
        return raw_phi, smoothed_phi, effective_phi

    def _sigmoid_alpha_controls(self, phi, previous_alpha_controls):
        raw_phi, smoothed_phi, phi_array = self._prepare_phi(
            phi,
            smoothing=self.phi_smoothing,
            dead_zone=self.phi_dead_zone,
            gain=1.0,
        )
        span = self.alpha_high - self.alpha_low
        logits = np.clip(-self.alpha_sharpness * phi_array, -60.0, 60.0)
        target_alpha = self.alpha_low + (span / (1.0 + np.exp(-logits)))
        previous_array = np.asarray(previous_alpha_controls, dtype=np.float64)
        momentum = min(max(self.alpha_momentum, 0.0), 0.999)
        alpha_controls = (momentum * previous_array) + ((1.0 - momentum) * target_alpha)
        alpha_controls = np.clip(alpha_controls, self.alpha_low, self.alpha_high)
        return {
            "controls": alpha_controls.tolist(),
            "target": target_alpha.tolist(),
            "raw_phi": raw_phi.tolist(),
            "smoothed_phi": smoothed_phi.tolist(),
            "effective_phi": phi_array.tolist(),
        }

    def _sigmoid_sigma_controls(self, phi, previous_sigma_controls):
        raw_phi, smoothed_phi, phi_array = self._prepare_phi(
            phi,
            smoothing=self.sigma_phi_smoothing,
            dead_zone=self.sigma_phi_dead_zone,
            gain=self.sigma_phi_gain,
        )
        span = self.sigma_high - self.sigma_low
        logits = np.clip(self.sigma_sharpness * phi_array, -60.0, 60.0)
        sigmoid_target = self.sigma_low + (span / (1.0 + np.exp(logits)))
        base_radius = self._base_radius_schedule()
        target_sigma = base_radius + (self.sigma_control_strength * (sigmoid_target - base_radius))
        target_sigma = np.clip(target_sigma, self.sigma_low, self.sigma_high)
        previous_array = np.asarray(previous_sigma_controls, dtype=np.float64)
        if previous_array.size != self.epochs:
            previous_array = base_radius.copy()
        momentum = min(max(self.sigma_momentum, 0.0), 0.999)
        max_step = max(self.sigma_step_clip, 0.0)
        sigma_controls = np.empty_like(target_sigma)
        sigma_deltas = np.empty_like(target_sigma)
        proposed = (momentum * previous_array[0]) + ((1.0 - momentum) * target_sigma[0])
        delta = np.clip(proposed - previous_array[0], -max_step, max_step)
        sigma_controls[0] = np.clip(previous_array[0] + delta, self.sigma_low, self.sigma_high)
        sigma_deltas[0] = sigma_controls[0] - previous_array[0]
        for t in range(1, len(target_sigma)):
            previous_sigma = sigma_controls[t - 1]
            proposed = (momentum * previous_sigma) + ((1.0 - momentum) * target_sigma[t])
            delta = np.clip(proposed - previous_sigma, -max_step, max_step)
            sigma_controls[t] = np.clip(previous_sigma + delta, self.sigma_low, self.sigma_high)
            sigma_deltas[t] = sigma_controls[t] - previous_sigma
        return {
            "controls": sigma_controls.tolist(),
            "target": target_sigma.tolist(),
            "raw_phi": raw_phi.tolist(),
            "smoothed_phi": smoothed_phi.tolist(),
            "effective_phi": phi_array.tolist(),
            "delta": sigma_deltas.tolist(),
        }

    def _base_radius_schedule(self):
        return np.asarray([sigma_t(self.sigma_high, t, self.epochs) for t in range(self.epochs)], dtype=np.float64)

    def _optimize_controls(self, forward_fn, backward_fn, backward_arg_builder, data, sample_indices, initial_weights):
        alpha_controls = np.full(self.epochs, self.alpha_high, dtype=np.float64)
        sigma_controls = self._base_radius_schedule()
        alpha_result = self._sigmoid_alpha_controls(np.zeros(self.epochs, dtype=np.float64), alpha_controls)
        sigma_result = self._sigmoid_sigma_controls(np.zeros(self.epochs, dtype=np.float64), sigma_controls)
        for _ in range(max(self.policy_passes, 1)):
            forward_outputs = forward_fn(data, initial_weights, sample_indices, alpha_controls.tolist(), sigma_controls.tolist())
            backward_inputs = backward_arg_builder(forward_outputs, alpha_controls.tolist(), sigma_controls.tolist())
            _, phi_alpha, phi_sigma = backward_fn(*backward_inputs)
            # Alpha and sigma are updated separately so the optimizer can decouple step size from neighborhood spread.
            # That extra degree of freedom often improves stability and MQE by avoiding the large-step, wide-neighborhood regime.
            alpha_result = self._sigmoid_alpha_controls(phi_alpha, alpha_controls)
            sigma_result = self._sigmoid_sigma_controls(phi_sigma, sigma_controls)
            alpha_controls = np.asarray(alpha_result["controls"], dtype=np.float64)
            sigma_controls = np.asarray(sigma_result["controls"], dtype=np.float64)

        forward_outputs = forward_fn(data, initial_weights, sample_indices, alpha_controls.tolist(), sigma_controls.tolist())
        backward_inputs = backward_arg_builder(forward_outputs, alpha_controls.tolist(), sigma_controls.tolist())
        lambdas, phi_alpha, phi_sigma = backward_fn(*backward_inputs)
        alpha_result = self._sigmoid_alpha_controls(phi_alpha, alpha_controls)
        sigma_result = self._sigmoid_sigma_controls(phi_sigma, sigma_controls)
        return (
            forward_outputs,
            alpha_controls.tolist(),
            sigma_controls.tolist(),
            alpha_result,
            sigma_result,
            lambdas,
        )

    def _radius_sensitivity(self, bmu, neighborhood_values, radius):
        radius = max(float(radius), 1e-12)
        neighborhood_array = np.asarray(neighborhood_values, dtype=np.float64)
        distance_sq = self._coord_distances[bmu] ** 2
        return neighborhood_array * (distance_sq / (radius ** 3))

    def _somoc1_sigma_projection(self, lambda_now, x_t, state_t, neighborhood_values, radius_sensitivity):
        local_deltas = x_t - state_t
        directional_alignment = np.einsum("d,nd->n", lambda_now, local_deltas)
        weighted_alignment = neighborhood_values * directional_alignment
        return float(np.sum(radius_sensitivity * weighted_alignment))

    def _forward_somoc1(self, data, initial_weights, sample_indices, alpha_controls, sigma_controls):
        data_array = _as_float_array(data)
        weights = _as_float_array(initial_weights).copy()
        states = [weights.copy()]
        bmus = []
        samples = []
        neighborhoods_history = []
        for t, sample_index in enumerate(sample_indices):
            x_t = data_array[sample_index]
            diff = weights - x_t
            bmu = int(np.argmin(np.einsum("ij,ij->i", diff, diff)))
            alpha_value = alpha_controls[t]
            radius = sigma_controls[t]
            # Alpha sets how far the state moves at each step, while sigma sets how broadly that movement spreads across the map.
            # Coupling them lets the controller use wide, gentle topology-forming steps early and narrow, stable refinements later.
            h_t = np.exp(-(self._coord_distances[bmu] ** 2) / (2.0 * max(radius, 1e-12) ** 2))
            weights += (alpha_value * h_t)[:, None] * (x_t - weights)
            states.append(weights.copy())
            bmus.append(bmu)
            samples.append(x_t.copy())
            neighborhoods_history.append(h_t.copy())
        state_metrics, gradients = _state_metrics(data_array, np.stack(states, axis=0))
        return states, bmus, samples, state_metrics[1:], neighborhoods_history, gradients[:-1]

    def _backward_somoc1(self, states, bmus, samples, neighborhoods_history, gradients, alpha_controls, sigma_controls):
        states_array = np.stack(states, axis=0)
        lambdas = np.zeros((self.epochs + 1, self.input_dim), dtype=np.float64)
        phi_alpha = np.zeros(self.epochs, dtype=np.float64)
        phi_sigma = np.zeros(self.epochs, dtype=np.float64)
        for t in range(self.epochs - 1, -1, -1):
            bmu = bmus[t]
            x_t = samples[t]
            neighborhood_values = neighborhoods_history[t]
            h_bmu = float(neighborhood_values[bmu])
            beta = alpha_controls[t] * h_bmu
            lambda_now = _backward_costate_step(lambdas[t + 1], gradients[t, bmu], beta)
            delta_bmu = x_t - states_array[t, bmu]
            projection = float(np.dot(lambda_now, delta_bmu))
            radius_sensitivity = self._radius_sensitivity(bmu, neighborhood_values, sigma_controls[t])
            sigma_projection = self._somoc1_sigma_projection(
                lambda_now,
                x_t,
                states_array[t],
                neighborhood_values,
                radius_sensitivity,
            )
            phi_alpha[t] = h_bmu * projection
            phi_sigma[t] = alpha_controls[t] * sigma_projection
            lambdas[t] = lambda_now
        return lambdas.tolist(), phi_alpha.tolist(), phi_sigma.tolist()

    def train_somoc1(self, data, sample_indices, initial_weights):
        start_time = time.perf_counter()
        (
            (states, bmus, samples, qe_history, neighborhoods_history, _),
            alpha_controls,
            sigma_controls,
            alpha_result,
            sigma_result,
            lambdas,
        ) = self._optimize_controls(
            self._forward_somoc1,
            self._backward_somoc1,
            lambda outputs, current_alpha_controls, current_sigma_controls: (
                outputs[0],
                outputs[1],
                outputs[2],
                outputs[4],
                outputs[5],
                current_alpha_controls,
                current_sigma_controls,
            ),
            data,
            sample_indices,
            initial_weights,
        )
        elapsed_seconds = time.perf_counter() - start_time
        return RunHistory(
            [state.tolist() for state in states],
            bmus,
            [sample.tolist() for sample in samples],
            qe_history.tolist(),
            alpha_controls,
            sigma_controls,
            sigma_controls,
            alpha_result["effective_phi"],
            sigma_result["effective_phi"],
            alpha_result["raw_phi"],
            sigma_result["raw_phi"],
            alpha_result["smoothed_phi"],
            sigma_result["smoothed_phi"],
            sigma_result["target"],
            sigma_result["delta"],
            lambdas,
            elapsed_seconds,
        )

    def _forward_somoc2(self, data, initial_weights, sample_indices, alpha_controls, sigma_controls):
        data_array = _as_float_array(data)
        weights = _as_float_array(initial_weights).copy()
        states = [weights.copy()]
        bmus = []
        samples = []
        neighborhoods_history = []
        for t, sample_index in enumerate(sample_indices):
            x_t = data_array[sample_index]
            diff = weights - x_t
            bmu = int(np.argmin(np.einsum("ij,ij->i", diff, diff)))
            alpha_value = alpha_controls[t]
            radius = sigma_controls[t]
            # Independent alpha and sigma controls improve stability because a large neighborhood can be paired with a small step size.
            # That reduces oscillation while still allowing the controller to reshape global topology before local convergence.
            h_t = np.exp(-(self._coord_distances[bmu] ** 2) / (2.0 * max(radius, 1e-12) ** 2))
            weights += (alpha_value * h_t)[:, None] * (x_t - weights)
            states.append(weights.copy())
            bmus.append(bmu)
            samples.append(x_t.copy())
            neighborhoods_history.append(h_t.copy())
        state_metrics, gradients = _state_metrics(data_array, np.stack(states, axis=0))
        return states, bmus, samples, state_metrics[1:], neighborhoods_history, gradients

    def _backward_somoc2(self, states, bmus, samples, neighborhoods_history, gradients, alpha_controls, sigma_controls):
        states_array = np.stack(states, axis=0)
        neighborhoods_array = np.stack(neighborhoods_history, axis=0)
        samples_array = np.stack(samples, axis=0)
        lambdas = np.zeros_like(states_array)
        phi_alpha = np.zeros(self.epochs, dtype=np.float64)
        phi_sigma = np.zeros(self.epochs, dtype=np.float64)
        for t in range(self.epochs - 1, -1, -1):
            beta = alpha_controls[t] * neighborhoods_array[t][:, None]
            lambda_now = _backward_costate_step(lambdas[t + 1], gradients[t], beta)
            x_t = samples_array[t]
            w_t = states_array[t]
            directional_term = np.einsum("ij,ij->i", lambda_now, x_t - w_t)
            radius_gradient = self._radius_sensitivity(bmus[t], neighborhoods_array[t], sigma_controls[t])
            phi_alpha[t] = float(np.sum(neighborhoods_array[t] * directional_term))
            phi_sigma[t] = float(alpha_controls[t] * np.sum(radius_gradient * directional_term))
            lambdas[t] = lambda_now
        return lambdas.tolist(), phi_alpha.tolist(), phi_sigma.tolist()

    def train_somoc2(self, data, sample_indices, initial_weights):
        start_time = time.perf_counter()
        (
            (states, bmus, samples, qe_history, _, _),
            alpha_controls,
            sigma_controls,
            alpha_result,
            sigma_result,
            lambdas,
        ) = self._optimize_controls(
            self._forward_somoc2,
            self._backward_somoc2,
            lambda outputs, current_alpha_controls, current_sigma_controls: (
                outputs[0],
                outputs[1],
                outputs[2],
                outputs[4],
                outputs[5][:-1],
                current_alpha_controls,
                current_sigma_controls,
            ),
            data,
            sample_indices,
            initial_weights,
        )
        elapsed_seconds = time.perf_counter() - start_time
        return RunHistory(
            [state.tolist() for state in states],
            bmus,
            [sample.tolist() for sample in samples],
            qe_history.tolist(),
            alpha_controls,
            sigma_controls,
            sigma_controls,
            alpha_result["effective_phi"],
            sigma_result["effective_phi"],
            alpha_result["raw_phi"],
            sigma_result["raw_phi"],
            alpha_result["smoothed_phi"],
            sigma_result["smoothed_phi"],
            sigma_result["target"],
            sigma_result["delta"],
            lambdas,
            elapsed_seconds,
        )


def summarize_history(history):
    return {
        "average_mqe": mean(history.quantization_error),
        "final_mqe": history.quantization_error[-1],
        "time_taken": history.elapsed_seconds,
    }


def run_single_benchmark(dataset_name, run_seed, k):
    data, labels = load_paper_dataset(dataset_name, normalize=True)
    config = PAPER_CONFIGS[dataset_name]
    rows, cols = config["map_size"]
    num_nodes = rows * cols
    input_dim = len(data[0])
    epochs = int(k)
    sample_rng = random.Random(run_seed)
    init_rng = random.Random(run_seed + 10000)
    sample_indices = [sample_rng.randrange(len(data)) for _ in range(epochs)]
    initial_weights = build_random_matrix(num_nodes, input_dim, init_rng)

    csom = ConventionalSOM(config["map_size"], input_dim, epochs=epochs, alpha0=0.5, sigma0=5.0)
    somoc = AdaptiveSOMOptimalControl(
        config["map_size"],
        input_dim,
        epochs=epochs,
        alpha_high=0.5,
        alpha_low=0.01,
        sigma_high=5.0,
        sigma_low=1.0,
        policy_passes=1,
    )

    csom_history = csom.train(data, sample_indices, initial_weights)
    somoc1_history = somoc.train_somoc1(data, sample_indices, initial_weights)
    somoc2_history = somoc.train_somoc2(data, sample_indices, initial_weights)

    return {
        "k": epochs,
        "CSOM": summarize_history(csom_history),
        "SOMOC1": summarize_history(somoc1_history),
        "SOMOC2": summarize_history(somoc2_history),
        "history": {
            "CSOM": csom_history,
            "SOMOC1": somoc1_history,
            "SOMOC2": somoc2_history,
        },
        "labels": labels,
        "data": data,
    }


def average_benchmark_results(dataset_name, runs=10, seed=7, k_values=None):
    tested_k_values = list(K_VALUES if k_values is None else k_values)
    results = {}
    for k in tested_k_values:
        aggregate = {
            algorithm: {"average_mqe": [], "final_mqe": [], "time_taken": []}
            for algorithm in ALGORITHMS
        }
        for run_index in range(runs):
            result = run_single_benchmark(dataset_name, run_seed=seed + run_index, k=k)
            for algorithm in ALGORITHMS:
                aggregate[algorithm]["average_mqe"].append(result[algorithm]["average_mqe"])
                aggregate[algorithm]["final_mqe"].append(result[algorithm]["final_mqe"])
                aggregate[algorithm]["time_taken"].append(result[algorithm]["time_taken"])
        results[k] = {}
        for algorithm in ALGORITHMS:
            results[k][algorithm] = {
                "average_mqe": mean(aggregate[algorithm]["average_mqe"]),
                "final_mqe": mean(aggregate[algorithm]["final_mqe"]),
                "time_taken": mean(aggregate[algorithm]["time_taken"]),
            }
    return results


def _metric_trend(values):
    if len(values) < 2:
        return "insufficient data"
    deltas = [next_value - current_value for current_value, next_value in zip(values, values[1:])]
    if all(delta <= 0.0 for delta in deltas) and any(delta < 0.0 for delta in deltas):
        return "decreasing"
    if all(delta >= 0.0 for delta in deltas) and any(delta > 0.0 for delta in deltas):
        return "increasing"
    if all(abs(delta) < 1e-12 for delta in deltas):
        return "flat"
    return "mixed"


def _infer_grid_shape(num_nodes):
    rows = int(math.sqrt(num_nodes))
    while rows > 1 and num_nodes % rows != 0:
        rows -= 1
    cols = num_nodes // rows
    return rows, cols


def _set_plot_style():
    try:
        import matplotlib.pyplot as plt
    except ModuleNotFoundError as exc:
        raise ModuleNotFoundError("matplotlib is required for plotting functions") from exc
    plt.rcParams["figure.dpi"] = 160
    plt.rcParams["savefig.dpi"] = 200
    return plt


def plot_mqe(history_dict):
    plt = _set_plot_style()
    plt.figure(figsize=(10, 6))
    for algorithm in ALGORITHMS:
        history = history_dict.get(algorithm)
        if history is None:
            continue
        epochs = np.arange(1, len(history.quantization_error) + 1)
        style = PLOT_STYLES[algorithm]
        plt.plot(
            epochs,
            history.quantization_error,
            label=algorithm,
            color=style["color"],
            linestyle=style["linestyle"],
            linewidth=2.0,
        )
    plt.title("MQE Over Training Epochs")
    plt.xlabel("Epoch")
    plt.ylabel("Mean Quantization Error (MQE)")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()


def plot_alpha_controls(history_dict):
    plt = _set_plot_style()
    plt.figure(figsize=(10, 6))
    for algorithm in ALGORITHMS:
        history = history_dict.get(algorithm)
        if history is None:
            continue
        epochs = np.arange(1, len(history.alpha_controls) + 1)
        style = PLOT_STYLES[algorithm]
        plt.plot(
            epochs,
            history.alpha_controls,
            label=algorithm,
            color=style["color"],
            linestyle=style["linestyle"],
            linewidth=2.0,
        )
    plt.title("Learning Rate Trajectories")
    plt.xlabel("Epoch")
    plt.ylabel("Learning Rate")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()


def plot_radius_controls(history_dict):
    plt = _set_plot_style()
    plt.figure(figsize=(10, 6))
    for algorithm in ALGORITHMS:
        history = history_dict.get(algorithm)
        if history is None:
            continue
        radius_values = history.radius_history if history.radius_history else history.sigma_controls
        epochs = np.arange(1, len(radius_values) + 1)
        style = PLOT_STYLES[algorithm]
        plt.plot(
            epochs,
            radius_values,
            label=algorithm,
            color=style["color"],
            linestyle=style["linestyle"],
            linewidth=2.0,
        )
    plt.title("Neighborhood Radius Trajectories")
    plt.xlabel("Epoch")
    plt.ylabel("Neighborhood Radius")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()


def plot_radius(history_dict):
    return plot_radius_controls(history_dict)


def plot_sigma_target_vs_actual(history_dict):
    plt = _set_plot_style()
    plt.figure(figsize=(10, 6))
    for algorithm in ("SOMOC1", "SOMOC2"):
        history = history_dict.get(algorithm)
        if history is None or not history.sigma_controls:
            continue
        epochs = np.arange(1, len(history.sigma_controls) + 1)
        style = PLOT_STYLES[algorithm]
        plt.plot(
            epochs,
            history.sigma_controls,
            label=f"{algorithm} sigma",
            color=style["color"],
            linestyle=style["linestyle"],
            linewidth=2.0,
        )
        plt.plot(
            epochs,
            history.sigma_target_history,
            label=f"{algorithm} sigma target",
            color=style["color"],
            linestyle=":",
            linewidth=1.6,
            alpha=0.9,
        )
    plt.title("Sigma Target vs Actual")
    plt.xlabel("Epoch")
    plt.ylabel("Sigma")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()


def plot_controls(history_dict):
    return plot_alpha_controls(history_dict)


def plot_switching_function_alpha(history_dict):
    plt = _set_plot_style()
    plt.figure(figsize=(10, 6))
    for algorithm in ("SOMOC1", "SOMOC2"):
        history = history_dict.get(algorithm)
        if history is None or not history.switching_function_alpha:
            continue
        epochs = np.arange(1, len(history.switching_function_alpha) + 1)
        style = PLOT_STYLES[algorithm]
        plt.plot(
            epochs,
            history.switching_function_alpha,
            label=algorithm,
            color=style["color"],
            linestyle=style["linestyle"],
            linewidth=2.0,
        )
    plt.title("Switching Function $\\phi_{\\alpha}$ Over Epochs")
    plt.xlabel("Epoch")
    plt.ylabel("$\\phi_{\\alpha}$")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()


def plot_sigma_phi_raw(history_dict):
    plt = _set_plot_style()
    plt.figure(figsize=(10, 6))
    for algorithm in ("SOMOC1", "SOMOC2"):
        history = history_dict.get(algorithm)
        if history is None or not history.raw_switching_function_sigma:
            continue
        epochs = np.arange(1, len(history.raw_switching_function_sigma) + 1)
        style = PLOT_STYLES[algorithm]
        plt.plot(
            epochs,
            history.raw_switching_function_sigma,
            label=algorithm,
            color=style["color"],
            linestyle=style["linestyle"],
            linewidth=2.0,
        )
    plt.title("Raw Switching Function $\\phi_{\\sigma}$")
    plt.xlabel("Epoch")
    plt.ylabel("Raw $\\phi_{\\sigma}$")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()


def plot_sigma_phi_smoothed(history_dict):
    plt = _set_plot_style()
    plt.figure(figsize=(10, 6))
    for algorithm in ("SOMOC1", "SOMOC2"):
        history = history_dict.get(algorithm)
        if history is None or not history.smoothed_switching_function_sigma:
            continue
        epochs = np.arange(1, len(history.smoothed_switching_function_sigma) + 1)
        style = PLOT_STYLES[algorithm]
        plt.plot(
            epochs,
            history.smoothed_switching_function_sigma,
            label=algorithm,
            color=style["color"],
            linestyle=style["linestyle"],
            linewidth=2.0,
        )
    plt.title("Smoothed Switching Function $\\phi_{\\sigma}$")
    plt.xlabel("Epoch")
    plt.ylabel("Smoothed $\\phi_{\\sigma}$")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()


def plot_switching_function_sigma(history_dict):
    plt = _set_plot_style()
    plt.figure(figsize=(10, 6))
    for algorithm in ("SOMOC1", "SOMOC2"):
        history = history_dict.get(algorithm)
        if history is None or not history.switching_function_sigma:
            continue
        epochs = np.arange(1, len(history.switching_function_sigma) + 1)
        style = PLOT_STYLES[algorithm]
        plt.plot(
            epochs,
            history.switching_function_sigma,
            label=algorithm,
            color=style["color"],
            linestyle=style["linestyle"],
            linewidth=2.0,
        )
    plt.title("Switching Function $\\phi_{\\sigma}$ Over Epochs")
    plt.xlabel("Epoch")
    plt.ylabel("$\\phi_{\\sigma}$")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()


def plot_switching_function(history_dict):
    return plot_switching_function_alpha(history_dict)


def plot_sigma_diagnostics(history_dict):
    plot_radius_controls(history_dict)
    plot_sigma_phi_raw(history_dict)
    plot_sigma_phi_smoothed(history_dict)
    plot_sigma_target_vs_actual(history_dict)


def plot_som_grid(data, weights, labels, title):
    plt = _set_plot_style()
    data_array = _as_float_array(data)
    weight_array = _as_float_array(weights)
    rows, cols = _infer_grid_shape(len(weight_array))

    plt.figure(figsize=(8, 7))
    # Plot the dataset as a faint background so the learned map can be compared to the data distribution.
    plt.scatter(
        data_array[:, 0],
        data_array[:, 1],
        c=labels if labels is not None else "#7f7f7f",
        cmap="tab20",
        s=18,
        alpha=0.22,
        edgecolors="none",
        label="Dataset points",
    )

    # Draw horizontal and vertical connections to reveal the learned SOM topology.
    for row in range(rows):
        start = row * cols
        stop = start + cols
        row_weights = weight_array[start:stop]
        plt.plot(row_weights[:, 0], row_weights[:, 1], color="#111111", linewidth=1.2, alpha=0.9)
    for col in range(cols):
        column_weights = weight_array[col::cols]
        plt.plot(column_weights[:, 0], column_weights[:, 1], color="#111111", linewidth=1.0, alpha=0.6)

    # Highlight the final node locations on top of the topology lines.
    plt.scatter(
        weight_array[:, 0],
        weight_array[:, 1],
        color="#ff7f0e",
        s=32,
        edgecolors="white",
        linewidths=0.5,
        label="SOM nodes",
        zorder=3,
    )
    plt.title(title)
    plt.xlabel("Feature 1")
    plt.ylabel("Feature 2")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()


def plot_k_comparison(results, metric):
    if metric not in {"average_mqe", "final_mqe"}:
        raise ValueError("metric must be 'average_mqe' or 'final_mqe'")

    plt = _set_plot_style()
    k_values = sorted(results.keys())
    plt.figure(figsize=(10, 6))
    for algorithm in ALGORITHMS:
        series = [results[k][algorithm][metric] for k in k_values]
        style = PLOT_STYLES[algorithm]
        plt.plot(
            k_values,
            series,
            label=algorithm,
            color=style["color"],
            linestyle=style["linestyle"],
            marker="o",
            linewidth=2.0,
            markersize=5,
        )
    metric_label = "Average MQE" if metric == "average_mqe" else "Final MQE"
    plt.title(f"{metric_label} vs k")
    plt.xlabel("k (epochs)")
    plt.ylabel(metric_label)
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()


def print_k_benchmark_comparison(dataset_name, runs=10, seed=7, k_values=None):
    tested_k_values = list(K_VALUES if k_values is None else k_values)
    results = average_benchmark_results(dataset_name, runs=runs, seed=seed, k_values=tested_k_values)
    print(f"Dataset: {dataset_name}")
    print("k | Algorithm | Avg MQE | Final MQE")
    for k in tested_k_values:
        for algorithm in ALGORITHMS:
            print(
                f"{k:4d} | "
                f"{algorithm:8s} | "
                f"{results[k][algorithm]['average_mqe']:.6f} | "
                f"{results[k][algorithm]['final_mqe']:.6f}"
            )
    print("Trend by algorithm:")
    for algorithm in ALGORITHMS:
        avg_series = [results[k][algorithm]["average_mqe"] for k in tested_k_values]
        final_series = [results[k][algorithm]["final_mqe"] for k in tested_k_values]
        print(
            f"{algorithm:8s} | "
            f"Avg MQE: {_metric_trend(avg_series)} | "
            f"Final MQE: {_metric_trend(final_series)}"
        )
    print()
    return results


if __name__ == "__main__":
    plt = _set_plot_style()
    aggregation_data, aggregation_labels = load_paper_dataset("aggregation")
    r15_data, r15_labels = load_paper_dataset("r15")

    print("Aggregation shape:", (len(aggregation_data), len(aggregation_data[0])), "labels:", len(aggregation_labels))
    print("R15 shape:", (len(r15_data), len(r15_data[0])), "labels:", len(r15_labels))
    print()

    aggregation_k_results = print_k_benchmark_comparison("aggregation", runs=10, seed=7, k_values=K_VALUES)
    r15_k_results = print_k_benchmark_comparison("r15", runs=10, seed=7, k_values=K_VALUES)

    # Example: visualize a single benchmark run using the stored training histories.
    result = run_single_benchmark("aggregation", run_seed=7, k=1000)
    history_dict = result["history"]
    plot_mqe(history_dict)
    plot_alpha_controls(history_dict)
    plot_radius_controls(history_dict)
    plot_sigma_target_vs_actual(history_dict)
    plot_sigma_phi_raw(history_dict)
    plot_sigma_phi_smoothed(history_dict)
    plot_switching_function_alpha(history_dict)
    plot_switching_function_sigma(history_dict)

    for algorithm in ALGORITHMS:
        plot_som_grid(
            result["data"],
            history_dict[algorithm].weights[-1],
            result["labels"],
            f"Final SOM Node Positions: {algorithm} on Aggregation",
        )

    plot_k_comparison(aggregation_k_results, metric="average_mqe")
    plot_k_comparison(aggregation_k_results, metric="final_mqe")
    plot_k_comparison(r15_k_results, metric="average_mqe")
    plot_k_comparison(r15_k_results, metric="final_mqe")
    plt.show()
```
